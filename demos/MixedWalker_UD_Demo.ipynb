# Mixed Walker UD Demo

Small baseline walk and utilization-distribution example using the current `MixedWalker` API.

```python
from pathlib import Path

from random_walk_package import MixedWalker, KernelMapping, TerrainMapHandle, plot_terrain_walk

repo_root = Path.cwd()
if repo_root.name == "demos":
    repo_root = repo_root.parent

terrain_path = repo_root / "random-walks" / "resources" / "random_walk_test_terrain_crop.txt"
terrain = TerrainMapHandle.from_file(str(terrain_path))

kernel_mapping = KernelMapping.mesa_default(terrain)
kernel_mapping.set_parameters(
    terrain=30,
    is_brownian=True,
    step_size=5,
    len_diffusity=0.8,
    angle_diffusity=0.1,
    max_bias_x=0,
    max_bias_y=0,
    directions=1,
)

steps = [(100, 100), (50, 200), (150, 200), (200, 300)]
T = 100
```

```python
plot_terrain_walk(terrain=terrain, steps=steps, title="Base Line")
```

```python
with MixedWalker(terrain, kernel_mapping) as walker:
    walk = walker.walk(steps=steps, T=T, plot=True, title="Random Walk")
    ud = walker.utilization_distribution(steps=steps, T=T, plot=True, title="Random Walk Utilization")

print("success")
```
